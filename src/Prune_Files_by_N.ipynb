We use `prune_by_ref.sh`

````python
from pimcscripts import pimchelp
import subprocess
import hg_utils
````

## We define all parameters of our simulation

````python
R = 2.9
cΔR = 0.0
L = 100 # Å
τ = 0.004 # 1/K
μ = -7.2 # K
sim_params = {'base_dir':hg_utils.base_dir(cΔR,cylinder=True,R=R,raw=True).replace('/MERGED',''),
              'L':L,'μ':μ, 'τ':τ}
````

## Get the list of estimator names

````python
cT = 0.5
est_names = pimchelp.get_file_list_from_params(**sim_params,T=cT,ftype='estimator')
````

## Go through all files and prune by $N$

````python
col_num = 8
N = 30
cwd = sim_params['base_dir']
for est_name in est_names:
    cmd = ['prune_by_ref.sh','-p',f'*{pimchelp.get_pimcid(est_name)}.dat','-r',est_name,'-c','8','-v',f'{N}', '-i'] #'-o','Test_Filtered']
    result = subprocess.run(cmd,capture_output=True,text=True,check=True,cwd=cwd)
    print(result.stdout, end="")
````
